# EDA - WC 2026 Football Prediction

Notebook này dùng để tái hiện nhanh phần exploratory data analysis cho báo cáo môn Data Science.

```python
import json
from pathlib import Path

import pandas as pd

project_root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
matches = pd.read_csv(project_root / 'data' / 'raw' / 'matches.csv', parse_dates=['date'])
matches.head()
```

```python
matches['result'] = matches.apply(lambda row: 'home_win' if row.home_score > row.away_score else ('away_win' if row.home_score < row.away_score else 'draw'), axis=1)
matches['total_goals'] = matches.home_score + matches.away_score
matches['result'].value_counts(normalize=True), matches['total_goals'].mean()
```
